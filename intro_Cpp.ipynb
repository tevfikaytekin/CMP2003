## A Note on IPython Magic Commands (like `%%writefile`)

Throughout this notebook, you have seen cells starting with `%%writefile`. It is extremely important to understand what these are:

* **Notebook-Only Feature:** Commands starting with `%` or `%%` are called **IPython Magic Commands**. They are special directives interpreted by the Jupyter/Colab notebook environment to perform helper tasks (like writing file contents, running shell scripts, or timing code).

* **Standard C++ Files:** A real C++ source file only contains the actual C++ code (starting with `#include <iostream>`, etc.), without any cell magic prefixes.

## 6. Pointers
A pointer is a variable that stores the memory address of another variable. They are very powerful in C++ for direct memory manipulation. You use `&` to get the address of a variable and `*` to dereference a pointer (get the value at that address).

````python
%%writefile pointers.cpp
#include <iostream>
using namespace std;

int main() {
    int score = 100;
    int* scorePointer = &score; // Stores the memory address of 'score'

    cout << "Value of score: " << score << endl;
    cout << "Memory address of score (&score): " << &score << endl;
    cout << "Value stored in scorePointer: " << scorePointer << endl;
    cout << "Value pointed to by scorePointer (*scorePointer): " << *scorePointer << endl;

    return 0;
}
````

````python
!g++ pointers.cpp -o pointers
!./pointers
````

### 6.1 The Power of Pointers (Why use them?)
Languages like Python and Java hide direct memory management from the programmer. While this makes them safer and easier to learn by preventing memory leaks and corruption, it restricts fine-grained control. C++ provides pointers for several powerful reasons:

1. **Performance and Efficiency:** Instead of copying large objects when passing them to functions, you can just pass a lightweight pointer (memory address).
2. **Direct Hardware Access:** In systems programming (like operating systems or embedded devices), you often need to read/write to specific physical memory addresses. Pointers allow this direct hardware manipulation.
3. **Dynamic Memory Allocation:** Pointers are essential for allocating memory on the fly (on the heap) when you don't know the required size at compile time (as we will see in Section 12).
4. **Pointer Arithmetic:** You can mathematically calculate and navigate through memory addresses directly.

Let's look at a classic example of **Pointer Arithmetic**. In C++, an array's name is actually just a pointer to its first element. By adding to the pointer, we can physically move through memory!

````python
%%writefile pointer_power.cpp
#include <iostream>
using namespace std;

int main() {
    // A simple list of 3 numbers
    int numbers[3] = {10, 20, 30};

    // The name 'numbers' acts as a pointer to the first element
    int* ptr = numbers;

    cout << "Using pointer arithmetic to navigate memory:" << endl;
    cout << "==========================================" << endl;

    cout << "Memory address in ptr: " << ptr << " | Value (*ptr): " << *ptr << endl;

    // Move the pointer to the next memory location (moves by the size of an 'int')
    ptr++;
    cout << "Memory address in ptr: " << ptr << " | Value (*ptr): " << *ptr << " (Moved to next element)" << endl;

    // Move it again
    ptr++;
    cout << "Memory address in ptr: " << ptr << " | Value (*ptr): " << *ptr << " (Moved to next element)" << endl;

    return 0;
}
````

## 8. Passing Parameters by Reference
By default, C++ passes arguments by value (making a copy). If you want a function to modify the original variable, you can pass it by reference using an ampersand (`&`) in the parameter list. This is also useful for performance when passing large objects, as it avoids making a copy.

````python
%%writefile pass_by_reference.cpp
#include <iostream>
using namespace std;

// Function taking parameters by reference
void swapNumbers(int &a, int &b) {
    int temp = a;
    a = b;
    b = temp;
}

int main() {
    int x = 10;
    int y = 20;

    cout << "Before swap: x = " << x << ", y = " << y << endl;

    // Call the function
    // x and y will be modified directly
    swapNumbers(x, y);

    cout << "After swap: x = " << x << ", y = " << y << endl;

    return 0;
}
````

````python
!g++ pass_by_reference.cpp -o pass_by_reference
!./pass_by_reference
````

## 9. Pass-by-Value vs. Pass-by-Reference
To clearly see the difference, let's look at a program that uses both methods to try and modify a variable.

* **Pass-by-Value (Default):** Passes a *copy* of the variable. Any changes made to the parameter inside the function are lost when the function ends, because they only apply to the local copy.
* **Pass-by-Reference (Using `&`):** Passes an *alias* to the actual variable. Any changes made inside the function directly affect the original variable passed in.

````python
%%writefile value_vs_reference.cpp
#include <iostream>
using namespace std;

// Pass-by-value: 'x' is a copy
void modifyByValue(int x) {
    x = 100;
}

// Pass-by-reference: 'x' is a reference to the original variable
void modifyByReference(int &x) {
    x = 100;
}

int main() {
    int val1 = 10;
    int val2 = 10;

    cout << "Initial values: val1 = " << val1 << ", val2 = " << val2 << endl;

    // Try to modify val1 by value
    modifyByValue(val1);
    cout << "After modifyByValue: val1 = " << val1 << " (Unchanged!)" << endl;

    // Try to modify val2 by reference
    modifyByReference(val2);
    cout << "After modifyByReference: val2 = " << val2 << " (Changed!)" << endl;

    return 0;
}
````

````python
!g++ value_vs_reference.cpp -o value_vs_reference
!./value_vs_reference
````

## 10. Inheritance (OOP)
Inheritance is a key feature of Object-Oriented Programming that allows you to create a new class (the **Derived Class** or Child) based on an existing class (the **Base Class** or Parent). This promotes code reusability.

The derived class inherits attributes and methods from the base class, and you can also add new attributes and methods to the derived class. In C++, we use a colon (`:`) to inherit from a class.

````python
%%writefile inheritance.cpp
#include <iostream>
#include <string>
using namespace std;

// Base class (Parent)
class Animal {
  public:
    string name;

    void eat() {
        cout << "I can eat!" << endl;
    }

    void sleep() {
        cout << "Zzz..." << endl;
    }
};

// Derived class (Child) inheriting from Animal
class Dog : public Animal {
  public:
    string breed;

    void bark() {
        cout << "Woof woof!" << endl;
    }
};

int main() {
    // Create an object of the Dog class
    Dog myDog;
    myDog.name = "Buddy";
    myDog.breed = "Golden Retriever";

    cout << "My dog's name is " << myDog.name << " and he is a " << myDog.breed << "." << endl;

    // Call methods inherited from the Animal class
    myDog.eat();
    myDog.sleep();

    // Call the method defined in the Dog class
    myDog.bark();

    return 0;
}
````

````python
# Compile and run the inheritance example
!g++ inheritance.cpp -o inheritance
!./inheritance
````

## 11. Dynamic Memory Allocation
In C++, you can allocate memory at runtime (dynamically) on the **heap** using the `new` operator. This is particularly useful when you don't know the exact size of an array or object in advance.

Whenever you dynamically allocate memory using `new`, you **must** release it when you are done using the `delete` operator. Failing to do so causes a **memory leak**, where the program keeps consuming memory unnecessarily.

*   `new`: Allocates memory and returns a pointer to it.
*   `delete`: Frees a single dynamically allocated variable.
*   `new[]` and `delete[]`: Used for allocating and freeing arrays.

````python
%%writefile dynamic_memory.cpp
#include <iostream>
using namespace std;

int main() {
    // 1. Dynamically allocating a single variable
    int* intPtr = new int;
    *intPtr = 42; // Assign a value

    cout << "Value of dynamically allocated int: " << *intPtr << endl;

    // Free the memory!
    delete intPtr;
    cout << "Memory for intPtr freed." << endl << endl;

    // 2. Dynamically allocating an array
    int size = 5;
    int* arrPtr = new int[size];

    cout << "Dynamically allocated array values: ";
    for (int i = 0; i < size; i++) {
        arrPtr[i] = (i + 1) * 10;
        cout << arrPtr[i] << " ";
    }
    cout << endl;

    // Free the memory for the array!
    delete[] arrPtr;
    cout << "Memory for array freed." << endl;

    return 0;
}
````

````python
# Compile and run the dynamic memory example
!g++ dynamic_memory.cpp -o dynamic_memory
!./dynamic_memory
````

## 12. Templates
Templates in C++ are a powerful tool for writing generic programs. They allow you to define functions or classes that can operate on different data types without having to rewrite the entire code for each type.

We define a template using the `template <typename T>` or `template <class T>` keyword, where `T` acts as a placeholder for the actual data type that will be used when the function or class is instantiated.

````python
%%writefile templates.cpp
#include <iostream>
#include <string>
using namespace std;

// A template function that returns the maximum of two values
template <typename T>
T findMax(T a, T b) {
    return (a > b) ? a : b;
}

int main() {
    // Testing with integers
    int int1 = 15, int2 = 25;
    cout << "Max of integers: " << findMax(int1, int2) << endl;

    // Testing with doubles
    double d1 = 3.14, d2 = 9.81;
    cout << "Max of doubles: " << findMax(d1, d2) << endl;

    // Testing with strings
    string s1 = "Apple", s2 = "Zebra";
    cout << "Max of strings: " << findMax(s1, s2) << endl;

    return 0;
}
````

````python
# Compile and run the templates example
!g++ templates.cpp -o templates
!./templates
````

## 12.1 Class Templates vs. Function Templates
While **Function Templates** allow you to write generic functions (like our `findMax` example), **Class Templates** allow you to write generic classes.

Here are the main differences:
1. **Purpose:** Function templates generate functions, while class templates generate classes.
2. **Type Deduction:** For function templates, the compiler can usually guess (deduce) the data type `T` from the arguments you pass (e.g., `findMax(10, 20)` automatically knows `T` is `int`). For class templates, you usually have to explicitly specify the data type when creating an object using angle brackets (e.g., `Box<int> myBox;`).

Let's look at an example of a Class Template.

````python
%%writefile class_templates.cpp
#include <iostream>
#include <string>
using namespace std;

// Class Template definition
template <class T>
class Box {
  private:
    T content;

  public:
    // Constructor
    Box(T val) {
        content = val;
    }

    // Method to get the content
    T getContent() {
        return content;
    }
};

int main() {
    // For class templates, we explicitly specify the type in angle brackets <>

    // Create a Box for an integer
    Box<int> intBox(100);
    cout << "Integer Box contains: " << intBox.getContent() << endl;

    // Create a Box for a string
    Box<string> stringBox("Hello World");
    cout << "String Box contains: " << stringBox.getContent() << endl;

    // Create a Box for a double
    Box<double> doubleBox(99.99);
    cout << "Double Box contains: " << doubleBox.getContent() << endl;

    return 0;
}
````

````python
# Compile and run the class templates example
!g++ class_templates.cpp -o class_templates
!./class_templates
````

## 13. Standard Template Library (STL)
The Standard Template Library (STL) is a powerful set of C++ template classes that provide general-purpose classes and functions. It saves you from reinventing the wheel!

The STL has three core components:
1.  **Containers:** Objects that store data collections (e.g., `vector`, `list`, `map`, `set`).
2.  **Algorithms:** Functions that perform operations on containers (e.g., `sort`, `search`, `copy`).
3.  **Iterators:** Objects that act like pointers, allowing you to traverse the elements of a container.

Let's look at an example using a `vector` (a dynamic array that can change size) and the `sort` algorithm.

````python
%%writefile stl_example.cpp
#include <iostream>
#include <vector>      // Include the vector container
#include <algorithm>   // Include algorithms like sort

using namespace std;

int main() {
    // 1. Container: Vector (dynamic array)
    // Unlike a standard array, a vector can grow or shrink in size
    vector<int> numbers = {5, 2, 8, 1, 9};

    cout << "Original vector: ";
    // Using a range-based for loop to iterate through the container
    for (int num : numbers) {
        cout << num << " ";
    }
    cout << endl;

    // Adding a new element to the end of the vector
    numbers.push_back(3);
    cout << "Added 3 to the vector." << endl;

    // 2 & 3. Algorithm & Iterators: Sort the vector
    // .begin() and .end() return iterators pointing to the start and end of the container
    sort(numbers.begin(), numbers.end());

    cout << "Sorted vector: ";
    for (int num : numbers) {
        cout << num << " ";
    }
    cout << endl;

    return 0;
}
````

````python
# Compile and run the STL example
!g++ stl_example.cpp -o stl_example
!./stl_example
````

## 14. File Read and Write Operations
In C++, file input and output (I/O) are handled by the `<fstream>` library.

There are three main classes included in this library:
*   `ofstream`: Used for **writing** to files.
*   `ifstream`: Used for **reading** from files.
*   `fstream`: Can be used for both reading and writing.

To use these classes, you need to include the `<fstream>` header. Let's look at a basic example where we write some text to a file named `example.txt` and then read it back line by line.

````python
%%writefile file_io.cpp
#include <iostream>
#include <fstream>
#include <string>
using namespace std;

int main() {
    // 1. Writing to a file
    // Create an ofstream object named outFile and open 'example.txt'
    ofstream outFile("example.txt");

    // Always check if the file was successfully opened
    if (outFile.is_open()) {
        outFile << "Hello, File!" << endl;
        outFile << "This is the second line of text written to the file." << endl;
        // Close the file when done to save changes and free resources
        outFile.close();
        cout << "Successfully wrote to the file 'example.txt'." << endl;
    } else {
        cout << "Unable to open file for writing." << endl;
    }

    // 2. Reading from a file
    string textLine;
    // Create an ifstream object named inFile to read from 'example.txt'
    ifstream inFile("example.txt");

    if (inFile.is_open()) {
        cout << "\n--- Reading from the file ---" << endl;
        // Read the file line by line using getline()
        while (getline(inFile, textLine)) {
            cout << textLine << endl; // Print each line to the console
        }
        inFile.close();
    } else {
        cout << "Unable to open file for reading." << endl;
    }

    return 0;
}
````

````python
# Compile and run the file I/O example
!g++ file_io.cpp -o file_io
!./file_io
````

## 16. Exception Handling
Exceptions provide a way to react to exceptional circumstances (like runtime errors) in programs by transferring control to special functions called handlers.

Exception handling in C++ is built upon three keywords:
*   `try`: Identifies a block of code for which particular exceptions will be activated.
*   `throw`: A program throws an exception when a problem shows up. This is done using a `throw` keyword.
*   `catch`: A program catches an exception with an exception handler at the place in a program where you want to handle the problem.

Let's look at an example that handles a potential division-by-zero error.

````python
%%writefile exceptions.cpp
#include <iostream>
using namespace std;

int main() {
    int numerator = 10;
    int denominator = 0;

    cout << "Attempting division..." << endl;

    try {
        // Check for the error condition before performing the operation
        if (denominator == 0) {
            throw "Division by zero error!"; // Throw a const char* exception
        }

        // This code won't execute if an exception is thrown
        int result = numerator / denominator;
        cout << "Result is: " << result << endl;
    }
    catch (const char* errorMessage) {
        // Catch block to handle the const char* exception
        cerr << "Caught an exception: " << errorMessage << endl;
    }

    cout << "Program continues executing normally after the try-catch block." << endl;

    return 0;
}
````

````python
# Compile and run the exceptions example
!g++ exceptions.cpp -o exceptions
!./exceptions
````

## 17. Performance Comparison: Python vs. C++

One of the main reasons developers choose C++ is its sheer execution speed. Since C++ is compiled directly into machine code, it runs significantly faster than interpreted languages like Python for CPU-intensive tasks.

Let's write a program in both languages that performs the exact same task: calculating the sum of all integers from `1` to `100,000,000` (100 million) and timing the execution.

````python
import time

print("--- Python Execution ---")
start_time = time.time()

total_sum = 0
for i in range(1, 100000001):
    total_sum += i

end_time = time.time()
python_duration = end_time - start_time

print(f"Sum: {total_sum}")
print(f"Time taken in Python: {python_duration:.6f} seconds")
````

````python
%%writefile speed_test.cpp
#include <iostream>
#include <chrono>

int main() {
    std::cout << "--- C++ Execution ---" << std::endl;
    auto start_time = std::chrono::high_resolution_clock::now();

    long long total_sum = 0;
    for (long long i = 1; i <= 100000000; ++i) {
        total_sum += i;
    }

    auto end_time = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double> duration = end_time - start_time;

    std::cout << "Sum: " << total_sum << std::endl;
    std::cout << "Time taken in C++: " << duration.count() << " seconds" << std::endl;

    return 0;
}
````

````python
# Compile C++ with optimizations turned on (-O3)
!g++ -O3 speed_test.cpp -o speed_test
!./speed_test
````
